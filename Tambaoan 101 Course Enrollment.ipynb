## IMPLEMENT THE RULES

```r
student_eligibility <- function(programming, math, attendance, prerequisite)
if (programming >= 80 && math >= 75 && attendance >= 80 && prerequisite == "Yes"){
  print("Eligible")
}else if (programming >= 75 && math >= 70 && attendance >= 75){
  print("Conditionally Eligible")
}else{
 print("Not Eligible")
}

student_eligibility(78,87,54,"No")

```

## APPLY THE FUNCTIONS TO THE STUDENTS

```r
classify_student <- function
(programming_grade, math_grade, attendance_percentage, has_prerequisite)
{
  if (programming_grade >= 80 &&
    math_grade >= 75 &&
    attendance_percentage >= 80 &&
    has_prerequisite == "Yes")
    {
    return("Eligible")
  } else if (programming_grade >= 75 &&
    math_grade >= 70 &&
    attendance_percentage >= 75)
    {
    return("Conditionally Eligible")
  } else {
    return("Not Eligible")
  }
}

# Assuming 'student' dataframe row and 'student_id' are available in the environment from previous cells.
student_classification <- classify_student(programming, math, attendance, student$Prerequisite)
cat("Student ID:", student_id, "is", student_classification, "\n")
```
